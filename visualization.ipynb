```python
import matplotlib.pyplot as plt

methods = ['ABMIL', 'CLAM_SB', 'TransMIL', 'DS_MIL']
flops_n = [1.264215424, 1.806223104, 7.220685312, 1.900224]
flops_k = [0.137665024, 0.196685824, 1.224774656, 0.20709888]

# Giảm kích thước hình (chiều rộng, chiều cao)
plt.figure(figsize=(3.0, 2.0), dpi=600) 

# Thu nhỏ nét vẽ và điểm đánh dấu một chút để hợp với hình nhỏ
plt.plot(methods, flops_n, marker='o', linewidth=2.0, markersize=6, label='N patches')
plt.plot(methods, flops_k, marker='s', linewidth=2.0, markersize=6, label='K patches ($K_p$=300)')

ymax = max(flops_n + flops_k)
plt.ylim(0, ymax + 0.6)

# Thu nhỏ cỡ chữ một chút để tránh bị lấn lướt trên hình nhỏ
plt.ylabel('FLOPs (G)', fontsize=6)
plt.xlabel('Method', fontsize=6)
plt.xticks(fontsize=6)
plt.yticks(fontsize=6)
plt.grid(axis='y', linestyle='--', alpha=0.35)
plt.legend(frameon=False, fontsize=5)

plt.tight_layout()
plt.savefig('flops_line_plot.pdf', bbox_inches='tight', dpi=600)
plt.show()
```
